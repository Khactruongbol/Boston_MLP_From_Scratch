# Lab 4 - Báo cáo mô hình dự đoán giá nhà Boston Housing

Notebook này là báo cáo tổng hợp theo phong cách Lab 2, Lab 3. Notebook chỉ dùng Markdown và hình ảnh sau khi train/tune model, không có code cell.

## 1. Define problem - Định nghĩa bài toán

Bài toán là dự đoán giá nhà trung vị `MEDV` trong Boston Housing. Đây là supervised regression vì target là giá trị liên tục.

Mục tiêu: cào data, lọc data sạch, chia feature, train model, tune model, test model, chọn best model và dùng model đó trong giao diện Python.

## 2. Hướng giải quyết

Pipeline gồm: cào raw data từ StatLib, reconstruct data, lọc data, chia feature/target, train baseline, tune hyperparameter trên train set, evaluate trên held-out test set, lưu best model và sinh hình ảnh báo cáo.

Các model được so sánh: MLP Regression, MLP Tuned, Linear Regression, Decision Tree Regression, Decision Tree Tuned, Random Forest Regression, Random Forest Tuned.

## 3. Sử dụng những gì

Chương trình sử dụng `pandas`, `numpy`, `scikit-learn`, `tensorflow-cpu`, `matplotlib`, `joblib`, `streamlit`, `pytest`. Sklearn model được đặt trong `Pipeline(StandardScaler(), model)` khi tune để tránh data leakage.

## 4. Data lấy từ đâu và cách lấy data

Nguồn data: Boston Housing raw data từ StatLib/CMU.

`http://lib.stat.cmu.edu/datasets/boston`

Chương trình cào raw text và lưu vào `data/raw/boston_raw.txt`. Không dùng `sklearn.datasets.load_boston()`.

## 5. Cách lọc data sạch

Raw data được bỏ phần mô tả, ghép mỗi mẫu từ 2 dòng, chuyển numeric, xóa duplicate, xóa missing value, kiểm tra schema và target `MEDV` dương. Sau khi lọc, data có 506 dòng và 14 cột.

## 6. Feature và target

Số lượng feature: **13**.

Feature: `CRIM`, `ZN`, `INDUS`, `CHAS`, `NOX`, `RM`, `AGE`, `DIS`, `RAD`, `TAX`, `PTRATIO`, `B`, `LSTAT`.

Target: `MEDV`, giá nhà trung vị theo nghìn USD.

## 7. Model optimization - Tối ưu hiệu năng model

Tuning chỉ dùng train set: Decision Tree và Random Forest dùng `GridSearchCV`, MLP Tuned dùng train-validation split và early stopping. Test set chỉ được dùng để đánh giá cuối cùng.

Kết quả: No tuned candidate improved over the Random Forest baseline; retained the strongest held-out test model.
